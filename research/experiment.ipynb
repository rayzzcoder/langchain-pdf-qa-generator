```python
file_path = ("data/kb-100-disaggregating-the-graduate-unemployment-in-pakistan.pdf")
loader = PyPDFLoader(file_path)
data = loader.load()
```

```python
from langchain_text_splitters import TokenTextSplitter

splitter_ques_gen = TokenTextSplitter(
    encoding_name="cl100k_base", #we did not have openai and this is a recognized one for tiktoken
    chunk_size=10000,
    chunk_overlap=200,
)
```

```python
chunk_ques_gen = splitter_ques_gen.split_text(question_gen)
```

```python
#from langchain_community.docstore.document import Document #legacy approach
from langchain_core.documents import Document

document_ques_gen = [Document(page_content = t) for t in chunk_ques_gen]
```

```python
splitter_ans_gen = TokenTextSplitter(
    encoding_name="cl100k_base",
    chunk_size=1500,
    chunk_overlap=100,
)
```

```python
document_ans_gen = splitter_ans_gen.split_documents(document_ques_gen)
```

```python
document_ans_gen
```

```python
len(document_ans_gen)
```

```python
from langchain_groq import ChatGroq

llm_ques_gen_pipeline = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0.3,
)
```

```python
prompt_template = """
You are an expert at creating questions based on social sciences material and documentation.
Your goal is to prepare a researcher for their exam and interview questions.
You will do this by asking questions about the text below:

------------
{text}
------------

Create 10 high-quality questions from the provided text.

Focus on important concepts, facts, findings, statistics, arguments,
and conclusions.

Make sure not to lose any important information. Do not generate unnecessary or repetitive questions.

Questions:
"""
```

```python
from langchain_core.prompts import PromptTemplate

prompt_questions = PromptTemplate(template=prompt_template,input_variables=['text'])
```

```python
refine_template = """
You are an expert at creating practice questions based on social sciences material and documentation.

Your task is to maintain and improve an existing list of practice questions.

Here is the CURRENT QUESTION LIST:
----------------
{existing_answers}
----------------

Here is NEW CONTEXT from the document:
----------------
{text}
----------------

Instructions:
1. Review the current question list against the new context.
2. Keep questions that already cover important information.
3. Improve an existing question only when the new context provides important additional information.
4. Add new questions only when the new context contains important information that is not already covered.
5. Remove or combine redundant questions.
6. The final output MUST contain 10 questions.
7. Do NOT ask the user for the existing questions.
8. Do NOT explain your process.
9. Output ONLY the numbered questions.

Final questions:
"""
```

```python
Refine_Prompt_Questions = PromptTemplate(
    input_variables=["existing_answers","text"],
    template=refine_template,
)
```

```python
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
```

```python
ques_gen_chain = (
    prompt_questions
    | llm_ques_gen_pipeline
    | output_parser
)

refine_ques_gen_chain = (
    Refine_Prompt_Questions
    | llm_ques_gen_pipeline
    | output_parser
)
```

```python
questions = ques_gen_chain.invoke({
    "text":document_ans_gen[0].page_content
})

for doc in document_ans_gen[1:]:
    questions = refine_ques_gen_chain.invoke({
        "existing_answers":questions,
        "text":doc.page_content
    })
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
embeddings = GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-2")
```

```python
#from langchain_classic.vectorstores import FAISS #legacy approach
from langchain_community.vectorstores import FAISS
```

```python
vector_store = FAISS.from_documents(document_ans_gen, embeddings)
```

```python
llm_ans_gen = ChatGroq(temperature=0.1,model="openai/gpt-oss-120b")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
```

```python
retriever = vector_store.as_retriever(search_kwargs={"k":4})
```

```python
def format_docs(docs):
    return "\n\n".join(d.page_content for d in docs)
```

```python
answer_prompt = ChatPromptTemplate.from_template(
    """Answer the question using only the context below.
If the context doesn't contain the answer, say so.

Context:
{context}

Question: {question}

Answer:"""
)
```

```python
answer_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | answer_prompt
    | llm_ans_gen
    | StrOutputParser()
)
```

```python
from pathlib import Path

output_path = Path("output/questions_and_answers.txt")
output_path.parent.mkdir(exist_ok=True)

questions_list = [q.strip() for q in questions.split("\n") if q.strip()]

with open(output_path, "w", encoding="utf-8") as f:
    for i, q in enumerate(questions_list, start=1):
        a = answer_chain.invoke(q)

        print(f"Q{i}:", q)
        print(f"A{i}:", a)
        print()

        f.write(f"Q{i}: {q}\n")
        f.write(f"A{i}: {a}\n")
        f.write("-" * 60 + "\n\n")
        f.flush()   # save to disk right away
```
